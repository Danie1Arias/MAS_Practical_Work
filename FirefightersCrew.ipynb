The fire personnel crew is critical for the Multy-Agent System as they are responisble for extinguishing the fire, mitigating its spread, and preventing escalation.

| **Role** | **Goal** | **Backstory** | **Allow Delegation** | **Tools** |

```python
firefighter_team_supervisor = Agent(
    role='Head of the firefighter team',
    goal='Assesses the situation upon arrival and establishes a plan of action',
    backstory=(
        "The Firefighter Team Supervisor is responsible for creating an action plan and delegating tasks to other firefighters. "
        "Thanks to his extensive experience and knowledge of different types of fires, he can establish an effective action plan"
    ),
    allow_delegation=False,
    verbose=True,
    llm='ollama/llama3.1'
)

firefighter_unit_class_a = Agent(
    role='Firefighting unit',
    goal='Extinguishing Class A fires or perform rescue tasks',
    backstory=(
        "Firefighter Unit Class A is a unit specialized in structural fires involving common fuels such as plastic, wood, brush, paper or rubber. "
        "This type of fire can be extinguished with water but, to improve autonomy, requires the use of a specific foam."
    ),
    allow_delegation=False,
    verbose=True,
    llm='ollama/llama3.1'
)

firefighter_unit_class_b = Agent(
    role='Firefighting unit',
    goal='Extinguishing Class B fires or perform rescue tasks',
    backstory=(
        "Firefighter Unit Class B is a unit specialized in fires of flammable gases such as ethanol, propane, butane, acetylene, hydrogen, methane."
    ),
    allow_delegation=False,
    verbose=True,
    llm='ollama/llama3.1'
)

firefighter_unit_class_c = Agent(
    role='Firefighting unit',
    goal='Extinguishing Class C fires or perform rescue tasks',
    backstory=(
        "Firefighter Unit Class C is a unit specialized in electrical fires. This unit uses Class D fire extinguishers, which are specifically designed "
        "for metal fires."
    ),
    allow_delegation=False,
    verbose=True,
    llm='ollama/llama3.1'
)
```

**Planning an extinction strategy**

| **Task Detail** | **Value** |
| --- | --- |
| **Description** | Assess the severity of the fire and evaluate the action to be taken in the following steps: (1) Rescue personnel in potential danger of death. (2) Assess the type of fire and activate the most appropriate unit. |
| **Agent** | `firefighter_team_supervisor` |
| **Async Execution** | False |
| **Expected Output** | Activate the proper unit to extinguish the fire and activate other units to perform rescue tasks |
| **Output Pydantic** | Rescue |
| **Max iterations** | 1 |


**Extinguish the fire**

| **Task Detail** | **Value** |
| --- | --- |
| **Description** | The most appropriate unit for extinguishing the fire must act to put it out. |
| **Agent** | `firefighter_unit_class_a` `firefighter_unit_class_b` `firefighter_unit_class_c` |
| **Async Execution** | True |
| **Expected Output** | Extinguish the fire (Active fire == False) |
| **Output Pydantic** | Rescue |
| **Max iterations** | 1 |


**Rescue the staff**

| **Task Detail** | **Value** |
| --- | --- |
| **Description** | Units not dedicated to extinguishing fires can carry out rescue tasks if necessary. |
| **Agent** | `firefighter_unit_class_a` `firefighter_unit_class_b` `firefighter_unit_class_c` |
| **Async Execution** | True |
| **Expected Output** | Reduce the number of people in danger due to the fire to 0 |
| **Output Pydantic** | Rescue |
| **Max iterations** | 1 |

```python
planning_extinction_strategy = Task(
    description='Assess the severity of the fire and evaluate the action to be taken in the following steps: '
    '(1) Rescue personnel in potential danger of death. (2) Assess the type of fire and activate the most appropriate unit.',
    agent=firefighter_team_supervisor,
    async_execution=False,
    expected_output=(
        "Activate the proper unit to extinguish the fire and activate other units to perform rescue tasks"
    ),
    output_pydantic=Rescue,
    max_iter=1
)

extinguish_fire = Task(
    description='The most appropriate unit for extinguishing the fire must act to put it out.',
    agent=[firefighter_unit_class_a, firefighter_unit_class_b, firefighter_unit_class_c],
    async_execution=True,
    expected_output=(
        "Extinguish the fire (Active fire == False)"
    )
)

rescue_staff = Task(
    description='A news article that has been spellchecked and is ready for publication. The article summary is provided by the writer agent',
    agent=[firefighter_unit_class_a, firefighter_unit_class_b, firefighter_unit_class_c],
    async_execution=True,
    expected_output=(
        "Reduce the number of people in danger due to the fire to 0"
    )
)
```
